```python
# 불용어 사전 모음

# 급여기준
w1 = {"기준", "산정", "급여", "방법", "치료", "기관", "진료", "충전", 
      "복합", "여가", "적용", "여부", "환자", "경우", "의뢰", 
      "비용", "재료", "인력", "사업", "관련", "협력", "관리", "동시", "동일", "전문", "의료", "인정"}

# 비급여 진료비
w2 = {"비급여", "비용", "보험", "차이", "적용", "회사", "풍", "확인", "체온계", "항목", "진료", "여가"}

# 심사/평가
w3 = {"평가", "적정", "대상", "방법", "주기", "기관", "급여", "기준", "청구", "지정", "심사", "결과"}

# 의료급여
w4 = {"의료", "급여", "입원", "수가", "절차", "기관", "대상자", "산정", "수", "진료", "선택"}

# 자동차보험
w5 = {"자동차", "보험", "환자", "방법", "기준", "관련", "교통사고", "인정", "병원"}

# 청구방법
w6 = {"청구", "방법", "번호", "오류", "요양", "관련", "제출", "통보", "산정", "비용"}
```

```python
import pandas as pd
from konlpy.tag import Komoran
from collections import Counter

# Komoran 로드
komoran = Komoran()

# 불용어 설정 
# stopwords = {"기준", "적용", "수가", "행위", "고시", "요양기관", "및", "등", "관련", }

# (행위) 포함된 행만 필터링
filtered_df = dfs[dfs['category'] == '급여기준']

# 전체 키워드 수집
words = []
for text in filtered_df["title"].dropna():
    nouns = komoran.nouns(str(text))
    filtered = [w for w in nouns if w not in w1 and len(w) > 1]
    words.extend(filtered)

# 카운트
counter = Counter(words)
df_act_keywords = pd.DataFrame(counter.items(), columns=["키워드", "빈도"]).sort_values(by="빈도", ascending=False)

# 결과 확인
print(df_act_keywords.head(20))
```
